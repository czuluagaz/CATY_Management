```python
def clean_data(df):
    # Replace gaps forward from the previous valid value in: 'tavg', 'tmax', 'tmin'
    df = df.fillna({'tavg': df['tavg'].ffill(), 'tmax': df['tmax'].ffill(), 'tmin': df['tmin'].ffill()})
    # Replace gaps forward from the previous valid value in: 'wdir', 'wspd', 'pres'
    df = df.fillna({'wdir': df['wdir'].ffill(), 'wspd': df['wspd'].ffill(), 'pres': df['pres'].ffill()})
    # Drop columns: 'snow', 'wpgt', 'tsun'
    df = df.drop(columns=['snow', 'wpgt', 'tsun'])
    return df

# Loaded variable 'df' from URI: c:\Users\zudel\projects\CATY_Management\DB\weather_beyne-heusay_2013-2024.csv
df = pd.read_csv(r'c:\Users\zudel\projects\CATY_Management\DB\weather_beyne-heusay_2013-2024.csv')

df_clean = clean_data(df.copy())
df_clean.head()
```
